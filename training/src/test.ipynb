```python
import re

def highlight_word_in_file(input_file, target_word,debut,fin):
    """
    Lit un fichier, ajoute des parenthèses autour de chaque occurrence
    du mot target_word, puis réécrit le contenu dans output_file.
    La recherche est insensible à la casse.
    """

    with open(input_file, "r", encoding="utf-8") as f:
        content = f.read()

    # Ajouter des parenthèses autour du mot exact (insensible à la casse)
    pattern = rf'\b({re.escape(target_word)})\b'
    highlighted_content = re.sub(pattern, rf'{debut}\1{fin}', content, flags=re.IGNORECASE)

    # Écrire le nouveau contenu dans le fichier de sortie
    return  highlighted_content
filename = "/Users/vanessaguerrier/Downloads/projet_TER_M2/data/GreenMIR/text_xml_nettoyer/article_1.txt"
start_tag = "<model>"
end_tag = "</model>"
word = "DiffTransfer"
    
extracted_texts = highlight_word_in_file(filename, word,start_tag,end_tag)
print(extracted_texts)
```
